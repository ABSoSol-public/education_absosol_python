# Control flow and functions
* now that we know datatypes, its time to control the flow of our program
* control flow means: decide what code runs, how often it runs, and group repeating instructions into reusable blocks
* in this lesson we cover conditional statements (if/elif/else), loops (while/for), comprehensions and functions (def)

## conditional statements
* a conditional statement checks a condition (something that results in True or False) and only runs the indented code block if the condition matches
* the code block after a conditional statement always starts with ":" and needs to be indented (like we learned in lesson 01)

```python
# comparison operators return a boolean (True/False)
print(42 == 42)   # equal
print(42 != 41)   # not equal
print(42 > 41)    # greater than
print(42 >= 42)   # greater or equal
print(42 < 100)   # smaller than
print(42 <= 42)   # smaller or equal

# logical operators combine multiple booleans
print(True and False)  # and --> both need to be True
print(True or False)   # or --> at least one needs to be True
print(not True)         # not --> flips the boolean
```

```python
# lets use comparison and logical operators inside a conditional statement
my_age = 17

if my_age >= 18:
    print("you are an adult")
elif my_age >= 13:
    print("you are a teenager")
else:
    print("you are a child")

# elif means "else if" - you can chain as many elif blocks as you want
# else is optional and only runs if none of the if/elif conditions matched
# python does not have a switch/case statement in older versions, elif-chains were the classic replacement
# since python 3.10 there is a structural pattern matching statement "match/case" which is closer to switch/case - we will see this in a later lesson
```

```python
# almost every value in python has a "truthy" or "falsy" interpretation, not only real booleans
# falsy values: False, None, 0, 0.0, "" (empty string), [] (empty list), {} (empty dict), set() (empty set)
# everything else is truthy

for value in [0, 1, "", "text", [], [1,2], None, {}, {"a":1}]:
    if value:
        print(value, "--> truthy")
    else:
        print(value, "--> falsy")

# this concept is heavily used in real python code, for example: "if my_list:" instead of "if len(my_list) > 0:"
```

```python
# python also has a compact one-line conditional expression, often called "ternary operator"
my_age = 20
status = "adult" if my_age >= 18 else "minor"
print(status)

# this is equivalent to a normal if/else but returns a value directly and can be assigned to a variable
# expert note: you can nest it, but readability suffers fast - use normal if/elif/else for more than one condition
```

## loops
* loops repeat a code block multiple times
* python has two loop types: "while" (repeat as long as a condition is True) and "for" (repeat over an iterable - a list, string, dict, range, ...)

```python
# while loop - repeats as long as the condition stays True
countdown = 5
while countdown > 0:
    print(countdown)
    countdown -= 1  # python has no ++ or -- operator like C/Java/JS, we use += / -= instead
print("liftoff!")

# careful: if the condition never becomes False you create an infinite loop
# "break" stops a loop immediately, "continue" skips to the next iteration
```

```python
# break and continue in action
number = 0
while True:  # an intentional infinite loop, only break gets us out
    number += 1
    if number % 2 == 0:
        continue  # skip even numbers, jump back to the while condition
    if number > 9:
        break  # stop the loop completely
    print(number)
```

```python
# for loop - iterates over an iterable, most common loop type in python
for character in "abc":
    print(character)

# range() generates a sequence of numbers, very often used together with for loops
# range(stop) --> 0 to stop-1
# range(start, stop) --> start to stop-1
# range(start, stop, step) --> start to stop-1 with stepping
for i in range(5):
    print("range(5):", i)

for i in range(2, 10, 2):
    print("range(2,10,2):", i)

# iterating over a list directly gives us the values, not the index
my_fruits = ["apple", "banana", "cherry"]
for fruit in my_fruits:
    print("fruit:", fruit)

# if we need both index and value, we use enumerate()
for index, fruit in enumerate(my_fruits):
    print(index, "-->", fruit)

# iterating over a dict by default gives us the keys, use .items() for key+value pairs
my_dict = {"a": 1, "b": 2, "c": 3}
for key in my_dict:
    print("key:", key)
for key, value in my_dict.items():
    print(key, "-->", value)
```

```python
# python loops have a special, rarely-seen-in-other-languages "else" clause
# the else block of a for/while loop runs only if the loop finished WITHOUT hitting a break
search_for = 99
my_numbers = [1, 5, 42, 7]

for number in my_numbers:
    if number == search_for:
        print("found it!")
        break
else:
    print("not found, loop finished completely without break")

# this pattern is very handy for search-loops, most other languages (JS, Java, C++) dont have this feature
# they need an extra boolean "found" flag variable instead
```

## comprehensions
* comprehensions are a very pythonic, compact way to build a new list/set/dict out of an existing iterable, combined with a loop and an optional condition in a single line
* they replace a lot of classic for-loop-and-append patterns

```python
# classic way with a normal for loop
squares_classic = []
for number in range(10):
    squares_classic.append(number ** 2)
print("classic:", squares_classic)

# the same result with a list comprehension: [expression for item in iterable]
squares_comprehension = [number ** 2 for number in range(10)]
print("comprehension:", squares_comprehension)

# comprehensions can also filter with an "if" at the end
even_squares = [number ** 2 for number in range(10) if number % 2 == 0]
print("even squares only:", even_squares)

# the same concept also works for dict and set
square_dict = {number: number ** 2 for number in range(5)}
print("dict comprehension:", square_dict)

square_set = {number ** 2 for number in [1, 1, 2, 2, 3]}
print("set comprehension (duplicates removed):", square_set)

# expert note: comprehensions are usually faster than the classic append-loop because the looping happens in optimized C code inside the interpreter
```

## functions
* a function is a named, reusable block of code
* functions help us to avoid duplicated code (DRY principle - "Dont Repeat Yourself") and to structure a program into small, testable pieces
* a function is defined with the keyword "def", a name, round brackets for parameters and a ":" to start the indented function body

```python
# a minimal function without parameters and without return
def say_hello():
    print("hello from inside a function")

say_hello()  # calling/executing the function

# a function with parameters
def greet(name):
    print(f"hello {name}")

greet("Alex")
greet("Mia")

# a function that returns a value with "return"
def add_numbers(a, b):
    return a + b

result = add_numbers(40, 2)
print("result:", result)

# without an explicit "return" a python function always returns "None" (the python equivalent of null/nil in other languages)
def no_return():
    print("i dont return anything")

print(no_return())
```

```python
# parameters can have default values, defaults must always come after non-default parameters
def greet_with_default(name, greeting="hello"):
    print(f"{greeting} {name}")

greet_with_default("Alex")              # uses the default greeting
greet_with_default("Alex", "good morning")  # overwrites the default

# parameters can also be called by keyword, in any order, if you name them
greet_with_default(greeting="hi", name="Mia")

# a function can also return multiple values at once - technically it returns one single tuple
def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 8, 15, 16, 23, 42])
print("lowest:", lowest, "highest:", highest)

# a docstring (like we learned in lesson 01) directly under the def line documents what a function does
def square(number):
    """Return the square of the given number."""
    return number ** 2

help(square)
```

## quick comparison to other languages
* python has no `switch`/`case` before version 3.10 (elif-chains are the classic replacement, `match`/`case` exists since 3.10 and is covered in a later lesson)
* python's `for item in iterable:` is closer to JavaScript's `for...of` or Java's `for (Type item : collection)` than to a classic C-style `for (int i=0; i<n; i++)` counting loop - if you need a counter, use `range()` or `enumerate()`
* python has no `++`/`--` operators, use `x += 1` / `x -= 1` instead
* python's `for...else` / `while...else` has no direct equivalent in JS, Java, C, C++, C# or Rust - those languages need a separate boolean flag variable to express "did the loop finish without a break"
* list comprehensions are similar in spirit to JavaScript's `array.map()`/`.filter()`, C#'s LINQ (`list.Select(...).Where(...)`) or Rust's iterator chains (`.map(...).filter(...)`), but comprehensions read left-to-right as one compact expression
* python functions are defined with `def name(params):`, compare to JavaScript `function name(params) { }` / arrow functions `(params) => { }`, Java/C++/C# `returnType name(params) { }`, Rust `fn name(params) -> ReturnType { }`
* a python function without `return` gives back `None` - the equivalent of JavaScript's `undefined`, Java/C#'s `null`, C/C++'s `NULL`/`nullptr`, or Rust's `()` (unit type)

```python
# sample solution 1 - fizzbuzz
def fizzbuzz(n):
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

fizzbuzz(15)

# sample solution 2 - is_prime using for...else
def is_prime(number):
    if number < 2:
        return False
    for divisor in range(2, number):
        if number % divisor == 0:
            return False
    else:
        return True

print([n for n in range(1, 20) if is_prime(n)])

# sample solution 3 - list comprehension
even_numbers = [n for n in range(1, 51) if n % 2 == 0]
print(even_numbers)

# sample solution 4 - count_vowels
def count_vowels(text):
    return sum(1 for character in text.lower() if character in "aeiou")

print(count_vowels("Python is awesome"))
```
